```python
#print FIZZ if a number is divisible by 3 , print BUZZ if a number is divisible by 5 , print FIZZBUZZ if it is divisible by both 3 and 5
a=int(input("Enter a number:"))
if(a%3==0 and a%5==0):
  print("FIZZBUZZ")
elif(a%3==0):
  print("FIZZ")
elif(a%5==0):
  print("BUZZ")
else:
  print("nothing is divisible")
```

```python
#print FIZZ if it is divisible by 3 only , print BUZZ if it is divisible by 5 only , print FIZZBUZZ if it is divisible by both
a=int(input("Enter a number:"))
if(a%3==0 and a%5!=0):
  print("FIZZ")
elif(a%3!=0 and a%5==0):
  print("BUZZ")
elif(a%3==0 and a%5==0):
  print("FIZZBUZZ")
else:
  print("nothing is divisible")
```

```python
#Programme on match
n=int(input("Enter a number:"))
match n:
  case 1:
    print("Sunday")
  case 2:
    print("Monday")
  case 3:
    print("Tuesday")
  case 4:
    print("Wednesday")
  case 5:
    print("Thursday")
  case 6:
    print("Friday")
  case 7:
    print("Saturday")
  case _:
    print("Invalid")
```

### LOOPS IN PYTHON

```python
#Printing from 1 to 10 numbers
i=1
while(i<=10):
  print(i)
  i=i+1
```

```python
#Printing from 10 to 1
i=10
while(i>=1):
  print(i)
  i=i-1
```

```python
#Printing even numbers from 1 to 10
i=1
while(i<=10):
  if(i%2==0):
    print(i)
  i=i+1
```

```python
#Printing even numbers from 10 to 1
i=10
while(i>=1):
  if(i%2==0):
    print(i)
  i=i-1
```
