```python
# 1. Creating a dictionary
d = {
    "name": "kishan",
    "age": 18,
    "grade": 8.67,
    "height": 5.75,
    "NAME":"sai"
}
print(d)
```

```python
d = {}
print(type(d))
```

```python
d1 = dict(a=35, n=14)
print(d1)
```

```python
for i in d:
    print(i)
```

```python
x = d.keys()
print(x)
print(type(x))
print(list(x))
```

```python
y = d.values()
print(y)
print(list(y))
```
