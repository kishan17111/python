```python
# Creating a file

f = open("demo.txt", "w")
f.write("Hello Python\n")
f.write("Welcome to File Handling")
f.close()

print("File created successfully")
```

```python
# Reading a file

f = open("demo.txt", "r")
print(f.read())
f.close()
```

```python
# Adding data to a file

f = open("demo.txt", "a")
f.write("\nThis is appended text")
f.close()

print("Data appended successfully")
```

```python
f = open("demo.txt", "r")
print(f.read(5))
f.close()
```

```python
with open("demo.txt", "r") as f:
    data = f.read()
    print("Number of characters:", len(data))
```
