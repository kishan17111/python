##Finding Prime Numbers

```python
n=int(input())
count=0
for i in range(1,n):
  if(n%i==0):
    count=1
if(count==0):
  print(n,"is prime")
else:
  print(n,"is not prime")
```

```python
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print(n,"is not a prime")
    break
else:
  print(n,"is a prime")
```

##Prime between Numbers

```python
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i)
```

```python
#sum of digits
n=int(input())
sum=0
while(n>0):
  r=n%10
  sum=sum+r
  n=n//10
print(sum)
```
