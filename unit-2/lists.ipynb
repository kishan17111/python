Operations on lists
1.Indexing
2.Slicing
3.Concatenation
4.Repetition
5.Membership Operator
6.Iteration

```python
#indexing
l=[36,7,"kishan",99.89,54]
print(l[2])
```

```python
#slicing
l=[36,7,"harsha","samar",75.36,27,189]
print(l[2:5])
print(l[0:7:2])
print(l[:5])
print(l[0:])
print(l[-1])
print([l[-5:-1]])
print(l[2:8:2])
print(l[::-1])
print(l[::-2])
```

```python
#concatenation
l1=[36,25,"python"]
l2=["java",76.45,136,99]
print(l1+l2)
```

```python
#repetition
print(l*3)
```

```python
#membership operator
print(5 in l)
```

```python
#iteration
for i in l:
  print(i,end=" ")
```
