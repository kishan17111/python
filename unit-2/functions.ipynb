```python
#palindrome
def palin(n):
  original=n
  rev=0
  while(n>0):
    r=n%10
    rev=rev*10+r
    n=n//10
  if rev==original:
    print("palindrome")
  else:
    print("not a palindrome")
n=int(input("enter a number"))
palin(n)
```

```python
#reversing a number
def rev(n):
  rev=0
  while(n>0):
    r=n%10
    rev=rev*10+r
    n=n//10
  return rev
n=int(input("enter a number"))
rev(n)
```

```python
#reversing a string
def rev_str(s):
  rev=s[::-1]
  return rev
s=input()
rev_str(s)
```

```python
def multiple_values(*nums):
  s=sum(nums)
  c=len(nums)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(3,5,7,13)
print(f"sum is {res_sum},count is {res_count},the average is {res_avg}")
```
