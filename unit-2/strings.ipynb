##String Methods

1.split
2.join
3.upper and lower
4.strip
5.replace
6.find
7.count

```python
text="perfect practice makes man perfect"
print(text.split())
print(text.split(","))
print(text.upper())
print(text.lower())
print(text.find("man"))
print(text.count("makes"))
print(text)
```
